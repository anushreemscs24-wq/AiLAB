```python

print("anushree m s 1WN24CS046")
import heapq

goal = [1, 2, 3, 8, 0, 4, 7, 6, 5]

# Calculate misplaced tiles
def h(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Print puzzle
def display(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


# A* Search
def astar(start):

    # (f, g, state, parent_path)
    queue = [(h(start), 0, start, [start])]
    visited = set()

    while queue:

        f, g, state, path = heapq.heappop(queue)

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        # Check goal
        if state == goal:
            for idx, s in enumerate(path):
                g_val = idx
                h_val = h(s)
                f_val = g_val + h_val
                print(f"Move: {g_val} | g(n) = {g_val}, h(n) = {h_val}, f(n) = {f_val}")
                display(s)
            print("Goal Reached!")
            return

        zero = state.index(0)

        row = zero // 3
        col = zero % 3

        # Find possible positions for 0
        positions = []

        if row > 0:
            positions.append(zero - 3)

        if row < 2:
            positions.append(zero + 3)

        if col > 0:
            positions.append(zero - 1)

        if col < 2:
            positions.append(zero + 1)

        # Create new states
        for pos in positions:

            new_state = state.copy()

            new_state[zero], new_state[pos] = \
                new_state[pos], new_state[zero]

            new_g = g + 1
            new_f = new_g + h(new_state)

            heapq.heappush(queue,
                           (new_f, new_g, new_state, path + [new_state]))


# Initial state
start = [2, 8, 3,
         1, 6, 4,
         0, 7, 5]


avoid_typo = astar(start)
```
